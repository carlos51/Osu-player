```python


data = []

for i in range(76):
    x = torch.load(f"tensors/rgb/batch_{i}.pt")
    #print(x.shape)
    data.append(x)

data = torch.cat(data,dim=0)

data.shape
```

```python
#last = y_raw[0]
datos_mouse = pd.read_csv("mouse/3_estrellas/bling_2.csv")
x_mouse = torch.tensor(datos_mouse["Mouse_X"],dtype= torch.float32).view(-1,1) / 1920
y_mouse = torch.tensor(datos_mouse["Mouse_Y"],dtype= torch.float32).view(-1,1) / 1080
click = torch.tensor(datos_mouse["Movimiento_Lateral"],dtype= torch.float32)
y_raw = torch.cat([x_mouse,y_mouse],dim=1)

y_raw.shape
```

```python
X = []
y = []


w = 20
n= len(click)

inicio = 386

k = inicio

for i in range(inicio,inicio + n - 32):
    X.append(data[i:i+32])
    #yy = torch.cat([x_mouse[i+32:i+32 + w],y_mouse[i+32:i+32 + w],click[i+32:i+32 + w]])
    y.append(y_raw[i+32:i+32 + w])

    if(len(y) == 32):
        print("k:",k)
        frames = torch.stack(X)
        targets = torch.stack(y)
        #print(frames.shape)
        #break
        torch.save({
            "X": frames,
            "y": targets
        }, f"tensors/3_estrellas/envy_baby/batch_{k}.pt")

        k += 1
        X = []
        y = []


```
